среднее арифметическое и медиана

```python
print("Среднее popularity:", df['popularity'].mean())
```

```python
print("Медиана popularity:", df['popularity'].median())
```

миниму,максимум,стандартное отклонение

```python
print("Минимум popularity:", df['popularity'].min())
print("Максимум popularity:", df['popularity'].max())
print("Стандартное отклонение popularity:", df['popularity'].std())
```

наиболее часто встречающийся жанр

```python
mode_genre = df['genre'].mode()
print("Мода для genre:")
print(mode_genre)
```

```python
mode_year = df['release_year'].mode()
print("Мода для release_year:")
print(mode_year)
```

гистограмма распределения popularity

```python
plt.figure(figsize=(10, 6))
plt.hist(df['popularity'], bins=20, color='steelblue', edgecolor='black')
plt.title('Распределение значений popularity', fontsize=14)
plt.xlabel('Popularity', fontsize=12)
plt.ylabel('Количество треков', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.show()
```
